```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
import math

xVal = np.linspace(0,0.9,200) 

u=(3.09)*10**(-4) # kg/m
T= 60 # N 
L=0.9 # m
A=0.005 #m
v=math.sqrt((T/u))

f=[]
w=[]
k=[]

for i in range(1,6):
    fs=(i/(2*L))*v
    f.append(fs)
    ws=2*(np.pi)*fs
    w.append(ws)
    ks=(2*np.pi)/(v/fs)
    k.append(ks)

def harmonics(x,t,n):
    wave=A*np.sin(k[n-1]*x+w[n-1]*t)+A*np.sin(k[n-1]*x-w[n-1]*t)
    
   
    return wave
    
#graph 
def pulse(t,n):
    plt.grid()
    plt.title(('Standing Wave of Harmonic',n,'Frequency: ',round(f[n-1],2),"Hz"))
    plt.ylabel('y')
    plt.xlabel('x')
    plt.xlim(0,0.9)
    plt.ylim(-2*A,2*A)
    plt.plot(xVal,harmonics(xVal,t,n))
    plt.show
#widgets in python are tools that enable interaction with the user

widgets.interact(pulse,t=(0,5,0.5),n=(1,5,1))

def summation(x,t):
    wave1=A*np.sin(k[0]*x+w[0]*t)+A*np.sin(k[0]*x-w[0]*t)
    wave2=A*np.sin(k[1]*x+w[1]*t)+A*np.sin(k[1]*x-w[1]*t)
    wave3=A*np.sin(k[2]*x+w[2]*t)+A*np.sin(k[2]*x-w[2]*t)
    wave4=A*np.sin(k[3]*x+w[3]*t)+A*np.sin(k[3]*x-w[3]*t)
    wave5=A*np.sin(k[4]*x+w[4]*t)+A*np.sin(k[4]*x-w[4]*t)
    waveSum=(0.001*wave1)+(0.004*wave2)+(0.002*wave3)+(0.04*wave4)+(0.002*wave5)
    return(waveSum)



#graph 
def pulse(t):
    plt.grid()
    plt.title('Standing Wave with all Harmonics')
    plt.ylabel('y')
    plt.xlabel('x')
    plt.xlim(0,0.9)
    plt.ylim(-0.1*A,0.1*A)
    plt.plot(xVal,summation(xVal,t))
    plt.show


widgets.interact(pulse,t=(0,5,0.005))



    
```
