```python
'''
111
'''
import vpython as vp
import matplotlib.pyplot as plt
import numpy as np

scene = vp.canvas()


m = 1.67*10**(-27) #kg
q = 1.6*10**(-19) #couloumb

Ex = 0
Ey = 0
Ez = 0
E = vp.vector(Ex, Ey, Ez) 

Bx = 0
By = 0
Bz = -0.5 #T
B = vp.vector(Bx, By, Bz)

vx = 6 * 10**7 #m/s
vy = 0
vz = 0
v = vp.vector(vx, vy,vz)

R = (m * v.mag)/(q * B.mag)

T = (2*np.pi * R)/(v.mag)
t = 0
tMax = T
dt = 0.001*tMax

x = 0
y = -R
z = 0

proton = vp.sphere(pos = vp.vector(x,y,z), radius = R, color = vp.color.yellow, make_trail = True)


while t < tMax:
    vp.rate(100)
    F = q* (E + vp.cross(v,B))
    a = F/m
    v = v + a*dt
    proton.pos = proton.pos + v*dt

    t = t+dt
    
```
