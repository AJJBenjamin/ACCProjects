```python
import numpy as np

#create the matrix
n = 75 # size of system

# Create a random matrix and make it diagonalizable
Q, _ = np.linalg.qr(np.random.randn(n, n))   # orthogonal matrix
eigenvalues = -np.linspace(1, 50, n)         # negative (stable system)
D = np.diag(eigenvalues)

A = Q @ D @ np.linalg.inv(Q)

#initial condition
x0 = np.random.randn(n)

# euler method coupled
def euler_coupled(A, x0, h, steps):
    x = x0.copy()
    for _ in range(steps):
        x = x + h * (A @ x)
    return x

#decouple the system
eigvals, P = np.linalg.eig(A)
P_inv = np.linalg.inv(P)

y0 = P_inv @ x0

#euler decoupled
def euler_decoupled(eigvals, y0, h, steps):
    y = y0.copy()
    for _ in range(steps):
        y = y + h * eigvals * y   # element-wise
    return y

# reverse transfrom
def reconstruct(P, y):
    return P @ y

#compare runtime
import time

h = 0.01
steps = 1000

# Coupled
start = time.time()
x_final = euler_coupled(A, x0, h, steps)
t1 = time.time() - start

# Decoupled
start = time.time()
y_final = euler_decoupled(eigvals, y0, h, steps)
x_final_decoupled = reconstruct(P, y_final)
t2 = time.time() - start

print("Coupled time:", t1)
print("Decoupled time:", t2)
print("Error:", np.linalg.norm(x_final - x_final_decoupled))

```
