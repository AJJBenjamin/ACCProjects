```python
import numpy as np
import pylab as pl


#create the grid 
w= 0.3
X = np.linspace(-w,w,100)
Y = np.linspace(-w,w,100)
Xm,Ym = np.meshgrid(X,Y)


# Variables
k = 9 * (10**9)
q = 3*(10**(-9))
d= 0.1

rp=((Xm**2)+((Ym-(d/2))**2))**(0.5)
rn=((Xm**2)+((Ym+(d/2))**2))**(0.5)


EpX = k*Xm*(q/rp**3)
EpY = k*(Ym-(0.5*d))*(q/rp**3)


EnX = k*Xm*(-q/rn**3)
EnY = k*(Ym+(0.5*d))*(-q/rn**3)

Ex=EpX+EnX
Ey=EpY+EnY


Vp = k*(q/rp)
Vn = k*(-q/rn)
Vnet= Vp+Vn
lines = [-200,-100,-50,50,100,200]


# Create the streamlines
pl.streamplot(Xm,Ym, Ex, Ey,3)
contours= pl.contour(Xm,Ym,Vnet,levels = lines, colors='black')
pl.clabel(contours, inline=True)
pl.plot(0,d/2,'ro') 
pl.plot(0,-d/2,'bo')

```
